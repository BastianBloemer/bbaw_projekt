```python
# Importe

import re
import json
import pandas as pd
import requests
import os

from IPython.display import display
pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', None)
```

```python
# Einlesen des Dataframes

df_all = pd.read_csv("../../data/curated/df_all.csv.csv")
```

```python
# PFADE
base_url = "https://digilib.bbaw.de/digilib/Manifester/IIIF/3/silo10!Bibliothek.tiff!{variable1}!{variable2}!tif"
unique_combinations = df_all[['schriftenreihe', 'band']].drop_duplicates()

# Links abrufen und JSONs speichern
for _, row in unique_combinations.iterrows():
    variable1 = row['schriftenreihe']
    variable2 = row['band']
    
    # ✅ output_dir HIER definieren, NACHDEM variable1 gesetzt wurde
    output_dir = f"../../data/raw/manifest/{variable1}"
    os.makedirs(output_dir, exist_ok=True)
    
    url = base_url.format(variable1=variable1, variable2=variable2)
    
    try:
        response = requests.get(url, timeout=10)
        response.raise_for_status()
        manifest_data = response.json()
        
        filename = f"{variable1}_{variable2}.json"
        filepath = os.path.join(output_dir, filename)
        
        with open(filepath, "w", encoding="utf-8") as f:
            json.dump(manifest_data, f, ensure_ascii=False, indent=2)
        
        print(f"✅ Gespeichert: {filepath}")
    
    except requests.exceptions.HTTPError as e:
        print(f"❌ HTTP-Fehler bei {url}: {e}")
    except requests.exceptions.ConnectionError:
        print(f"❌ Verbindungsfehler bei {url}")
    except requests.exceptions.Timeout:
        print(f"⏱️ Timeout bei {url}")
    except json.JSONDecodeError:
        print(f"⚠️ Kein gültiges JSON bei {url}")
```
